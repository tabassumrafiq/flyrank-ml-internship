## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

The goal is to classify whether a webpage needs a content refresh. Each webpage belongs to one of two categories:

Needs Refresh
Does Not Need Refresh

This is a binary classification problem because the output is one of two possible classes.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Needs_Refresh

Possible values:

1 = Needs Refresh
0 = Does Not Need Refresh

Where does the label come from?

The label can be created using observed performance indicators such as:

Declining organic traffic
Low click-through rate (CTR)
Poor search ranking
Old content that has not been updated recently

These indicators help determine whether a page should be refreshed.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metric: F1-Score

Why?

F1-Score balances Precision and Recall, making it a good choice when the number of pages needing refresh may be much smaller than the number of healthy pages.

Goal:

An F1-Score of 0.80 or higher would indicate strong model performance.Primary Metric: F1-Score

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content_refresh.csv")

# 1. My Lane as an ML Task
lane = "Content Refresh / SEO Performance"
task_type = "Classification"

# 2. Target or Proxy
target = "Needs_Refresh"

# 3. Success Metric
metric = "F1-Score"

# 4. Unit of Analysis
unit = "One row = One webpage"

print("Lane:", lane)
print("ML Task Type:", task_type)
print("Target:", target)
print("Success Metric:", metric)
print("Unit of Analysis:", unit)

print("\nDataset Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nTarget Distribution:")
print(df[target].value_counts())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*



A simple rule such as:

"Refresh every page older than 180 days."

does not consider many important factors.

Machine learning can learn from multiple features at the same time, including:

Organic traffic
Click-through rate (CTR)
Search ranking
Bounce rate
User engagement
Content age

Because these factors interact in complex ways, ML can make better predictions than a single fixed rule. This helps prioritize the pages that are most likely to benefit from a content refresh.
